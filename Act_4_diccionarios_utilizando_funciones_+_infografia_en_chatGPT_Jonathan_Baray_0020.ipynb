#### Farmaceutico

```python
# ============================================
# UIII ACT 4 - DICCIONARIOS UTILIZANDO FUNCIONES
# FARMACÉUTICO
# ============================================

# Diccionario vacío
diccionario_farmaceutico = []

# ============================================
# FUNCIÓN AGREGAR
# ============================================

def agregar_farmaceutico():
    print("\n--- AGREGAR FARMACÉUTICO ---")

    registro = input("Ingrese registro sanitario: ")
    farmacia = input("Ingrese nombre de farmacia: ")
    inventario = input("Ingrese nivel de inventario: ")
    especialidad = input("Ingrese especialidad en fórmulas: ")
    turnos = input("Ingrese turnos: ")
    ventas = input("Ingrese ventas del mes: ")

    farmaceutico = {
        "Registro": registro,
        "Farmacia": farmacia,
        "Inventario": inventario,
        "Especialidad": especialidad,
        "Turnos": turnos,
        "Ventas": ventas
    }

    diccionario_farmaceutico.append(farmaceutico)

    print("\nFarmacéutico agregado correctamente")
    listar_farmaceutico()

# ============================================
# FUNCIÓN LISTAR
# ============================================

def listar_farmaceutico():

    print("\n========== LISTA DE FARMACÉUTICOS ==========")

    print("--------------------------------------------------------------------------")
    print("Registro | Farmacia | Inventario | Especialidad | Turnos | Ventas")
    print("--------------------------------------------------------------------------")

    for datos in diccionario_farmaceutico:

        for key, value in datos.items():
            print(value, end=" | ")

        print()

    print("--------------------------------------------------------------------------")

# ============================================
# FUNCIÓN BUSCAR
# ============================================

def buscar_farmaceutico():

    print("\n--- BUSCAR FARMACÉUTICO ---")

    buscar = input("Ingrese el registro sanitario a buscar: ")

    for datos in diccionario_farmaceutico:

        if datos["Registro"] == buscar:

            print("\nFarmacéutico encontrado")
            print("-------------------------------------")

            for key, value in datos.items():
                print(key, ":", value)

            print("-------------------------------------")

# ============================================
# FUNCIÓN BORRAR
# ============================================

def borrar_farmaceutico():

    print("\n--- BORRAR FARMACÉUTICO ---")

    borrar = input("Ingrese el registro sanitario a borrar: ")

    for datos in diccionario_farmaceutico:

        if datos["Registro"] == borrar:
            diccionario_farmaceutico.remove(datos)

            print("\nFarmacéutico eliminado correctamente")

    listar_farmaceutico()

# ============================================
# FUNCIÓN ACTUALIZAR
# ============================================

def actualizar_farmaceutico():

    print("\n--- ACTUALIZAR FARMACÉUTICO ---")

    actualizar = input("Ingrese el registro sanitario a actualizar: ")

    for datos in diccionario_farmaceutico:

        if datos["Registro"] == actualizar:

            datos["Farmacia"] = input("Nueva farmacia: ")
            datos["Inventario"] = input("Nuevo nivel de inventario: ")
            datos["Especialidad"] = input("Nueva especialidad: ")
            datos["Turnos"] = input("Nuevo turno: ")
            datos["Ventas"] = input("Nuevas ventas del mes: ")

            print("\nInformación actualizada correctamente")

    listar_farmaceutico()

# ============================================
# DATOS INICIALES
# ============================================

diccionario_farmaceutico.append({
    "Registro": "RS101",
    "Farmacia": "Farmacia Central",
    "Inventario": "Alto",
    "Especialidad": "Fórmulas magistrales",
    "Turnos": "Matutino",
    "Ventas": "45000"
})

diccionario_farmaceutico.append({
    "Registro": "RS102",
    "Farmacia": "Farmacia Vida",
    "Inventario": "Medio",
    "Especialidad": "Medicamentos controlados",
    "Turnos": "Vespertino",
    "Ventas": "38000"
})

diccionario_farmaceutico.append({
    "Registro": "RS103",
    "Farmacia": "Farmacia Salud",
    "Inventario": "Bajo",
    "Especialidad": "Antibióticos",
    "Turnos": "Nocturno",
    "Ventas": "29000"
})

diccionario_farmaceutico.append({
    "Registro": "RS104",
    "Farmacia": "Farmacia Médica",
    "Inventario": "Alto",
    "Especialidad": "Vitaminas",
    "Turnos": "Mixto",
    "Ventas": "51000"
})

# ============================================
# MENÚ PRINCIPAL
# ============================================

while True:

    print("\n===================================")
    print(" SISTEMA DE FARMACÉUTICOS ")
    print("===================================")

    print("1. Agregar farmacéutico")
    print("2. Listar farmacéuticos")
    print("3. Buscar farmacéutico")
    print("4. Borrar farmacéutico")
    print("5. Actualizar farmacéutico")
    print("6. Salir")

    opcion = input("\nSeleccione una opción: ")

    if opcion == "1":
        agregar_farmaceutico()

    elif opcion == "2":
        listar_farmaceutico()

    elif opcion == "3":
        buscar_farmaceutico()

    elif opcion == "4":
        borrar_farmaceutico()

    elif opcion == "5":
        actualizar_farmaceutico()

    elif opcion == "6":
        print("\nPrograma finalizado")
        break

    else:
        print("\nOpción incorrecta")
```
